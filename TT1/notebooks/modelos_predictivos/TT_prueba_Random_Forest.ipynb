```python
# Importar las librerias
import pandas as pd
import os
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
from sklearn.preprocessing import LabelEncoder
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import sklearn
import matplotlib

print(f"Versión de Pandas: {pd.__version__}")
print(f"Versión de Scikit-learn: {sklearn.__version__}")
print(f"Versión de Matplotlib: {matplotlib.__version__}")
print(f"Versión de Seaborn: {sns.__version__}")
print(f"Versión de Joblib: {joblib.__version__}")
```

```python
# Cargar el archivo csv desde ruta local
file_path = os.path.join("..", "..", "data", "processed", "Dataset_Final_CE.csv")
df = pd.read_csv(file_path)
df.head()
```

```python
# Codificar columnas categoricas
label_encoders = {}
for col in df.select_dtypes(include='object').columns:
    le = LabelEncoder()
    df[col] = le.fit_transform(df[col])
    label_encoders[col] = le

# Definir atributos y clase
X = df.drop("Condition", axis=1)
y = df["Condition"]
print(label_encoders["Condition"].classes_)
```

```python
# Dividir en entrenamiento y prueba
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
```

```python
# Crear y entrenar el modelo
Modelo_RandomForest = RandomForestClassifier(n_estimators=357, max_depth=45,random_state=0, n_jobs=-1)
Modelo_RandomForest.fit(X_train, y_train)

y_pred = Modelo_RandomForest.predict(X_test)
```

```python
# Guardar el modelo y los encoders en archivos .pkl (ruta local)
model_path = "random_forest_model.pkl"
encoders_path = "random_forest_encoders.pkl"

# Guardar el modelo
joblib.dump(Modelo_RandomForest, model_path)
print(f"Modelo guardado en {model_path}")

# Guardar el diccionario de LabelEncoders
joblib.dump(label_encoders, encoders_path)
print(f"Label encoders guardados en {encoders_path}")
```

```python
# Mostrar resultados
print(label_encoders["Condition"].classes_)
sns.heatmap(confusion_matrix(y_test, y_pred), annot=True, fmt='d', cmap='Blues')
plt.xlabel("Predicted")
plt.ylabel("True")
plt.title("Matriz de Confusión - Random Forest")
plt.show()

print("Accuracy:", accuracy_score(y_test, y_pred))
print("\nClassification Report:\n", classification_report(y_test, y_pred))
```
